```python
# %% CELL 1 — Paths & Settings
DF_OUT_DIR   = './nuecc_dfs'
SYST_DIR     = './saved_syst'
PLOT_DIR     = './plots_xsec'
COV_OUT_DIR  = './nuecc_cov'
TARGET_POT   = 6.6e20
TITLE        = r'SBND $\nu_e$ CC Inclusive'
POT_LABEL    = r'$6.6\times10^{20}$ POT'
FINAL_STAGE  = 'sel_vertex_distance'
 
SEL_FILE     = f'{DF_OUT_DIR}/selected_nuecc_qual.df'
DF_FILE      = '/exp/sbnd/data/users/castalyf/nue_sel/production_files/gen1/mc1e20_nueCC_v2-2.df'
 
OMNIFOLD_DATA    = '/home/castalyf/FormattedData_SBND/'
OMNIFOLD_COV     = '/home/castalyf/Omnifold_SBND/sbnd/covariance/'
 
SAVE_FIGS = True
for d in [PLOT_DIR, COV_OUT_DIR, f'{PLOT_DIR}/fakedata',
          f'{PLOT_DIR}/omnifold_comparison']:
    os.makedirs(d, exist_ok=True)
 
def savefig(fname, fig=None):
    if SAVE_FIGS:
        if fig is None: fig = plt.gcf()
        fig.savefig(os.path.join(PLOT_DIR, fname), bbox_inches='tight', dpi=150)
 
print(f"Paths OK")
```

```python
# %% CELL 2 — Load sel_topo
st = nh.load_sel_topo(SEL_FILE, df_file=DF_FILE)
sel_topo      = st['sel_topo']
pot_scale     = st['pot_scale']
n_true_signal = st['n_true_signal']
 
sel_final = sel_topo[sel_topo[FINAL_STAGE]]
sig_mask  = (sel_final['truth_cat'] == 0)
bkg_mask  = sel_final['truth_cat'].between(1, 6)
 
print(f"\nsel_topo: {sel_final.shape}")
print(f"Signal: {sig_mask.sum():,}, Background: {bkg_mask.sum():,}")
print(f"Purity: {sig_mask.sum()/len(sel_final)*100:.1f}%")
```

```python
# %% CELL 3 — Variable configs
VAR_CFGS = {
    'reco_ke': VariableConfig(
        'reco_ke', SYST_KE_BINS,
        r'Reco leading-$e^-$ KE [MeV]', save_name='reco_ke'),
    'reco_costheta': VariableConfig(
        'reco_costheta', SYST_COS_BINS,
        r'Reco leading-$e^-$ $\cos\theta$', save_name='reco_costheta'),
}
TRUE_COL_MAP  = {'reco_ke': 'true_ke', 'reco_costheta': 'true_costheta'}
TRUE_BINS_MAP = {'reco_ke': SYST_KE_BINS, 'reco_costheta': SYST_COS_BINS}
 
for name, vcfg in VAR_CFGS.items():
    print(f"  {name}: {vcfg.n_bins} bins")
```

```python
# %% CELL 4 — Response matrix + efficiency  [FIXED]
# ══════════════════════════════════════════════════════════════════════════════
# Uses cafpyana's get_smear_matrix / get_response_matrix to ensure correct
# matrix orientation: Response has shape (n_reco × n_true).
 
_eps = 1e-8
response_dict = {}
 
for var_name, vcfg in VAR_CFGS.items():
    tc = TRUE_COL_MAP[var_name]
    tb = TRUE_BINS_MAP[var_name]
    bins = vcfg.bins
 
    sel = sel_final.copy()
 
    # ── Selected signal: reco + truth values ──────────────────────────────
    sig_sel = sel[sig_mask].copy()
    bkg_sel = sel[bkg_mask].copy()
 
    sig_reco = sig_sel[var_name].dropna().clip(bins[0], bins[-1] - _eps).values
    sig_true = sig_sel[tc].dropna().clip(tb[0], tb[-1] - _eps).values
 
    # Align lengths (both reco and true must be valid for same events)
    valid_both = sig_sel[var_name].notna() & sig_sel[tc].notna()
    sig_reco_aligned = sig_sel.loc[valid_both, var_name].clip(bins[0], bins[-1]-_eps).values
    sig_true_aligned = sig_sel.loc[valid_both, tc].clip(tb[0], tb[-1]-_eps).values
 
    # ── CV histograms (POT-weighted) ─────────────────────────────────────
    sig_reco_cv, _ = np.histogram(sig_reco, bins=bins,
                                   weights=np.full(len(sig_reco), pot_scale))
    bkg_reco_cv, _ = np.histogram(
        bkg_sel[var_name].dropna().clip(bins[0], bins[-1]-_eps),
        bins=bins, weights=np.full(bkg_mask.sum(), pot_scale))
 
    # ── Truth histogram: ALL signal (efficiency denominator), UNWEIGHTED ──
    all_sig = sel_topo[sel_topo['is_sig']]
    all_true = all_sig[tc].dropna().clip(tb[0], tb[-1]-_eps).values
    sig_true_all, _ = np.histogram(all_true, bins=tb)  # RAW counts, no pot_scale
 
    # ── Smearing matrix using cafpyana convention ─────────────────────────
    # get_smear_matrix returns reco_vs_true with shape (n_true × n_reco)
    # but internally does Response = reco_vs_true.T → (n_reco × n_true)
    reco_vs_true = get_smear_matrix(
        sig_true_aligned, sig_reco_aligned, bins,
        var_labels=[vcfg.xlabel, vcfg.xlabel, vcfg.xlabel_true],
        plot=False)
    # reco_vs_true shape: (n_true × n_reco)  [raw 2D histogram]
 
    # ── Efficiency ────────────────────────────────────────────────────────
    eff = get_eff(reco_vs_true, sig_true_all.astype(float))
 
    # ── Response matrix (n_reco × n_true) ─────────────────────────────────
    Response = get_response_matrix(
        reco_vs_true, eff, bins,
        var_labels=[vcfg.xlabel, vcfg.xlabel, vcfg.xlabel_true],
        plot=False)
    # Response shape: (n_reco × n_true)  ← CORRECT for WienerSVD
 
    response_dict[var_name] = dict(
        response=Response,
        reco_vs_true=reco_vs_true,
        efficiency=eff,
        sig_reco_cv=sig_reco_cv,
        bkg_reco_cv=bkg_reco_cv,
        sig_true_cv=sig_true_all.astype(float),  # unweighted
    )
 
    print(f"\n{var_name}:")
    print(f"  Response shape: {Response.shape}  (should be n_reco × n_true)")
    print(f"  Response diag:  {np.diag(Response)}")
    print(f"  Response diag mean: {np.mean(np.diag(Response)):.4f}")
    print(f"  Response col sums: {Response.sum(axis=0)}")
    print(f"  Efficiency: {np.round(eff, 3)}")
    print(f"  Sig reco CV (weighted): {sig_reco_cv.sum():.1f}")
    print(f"  Bkg reco CV (weighted): {bkg_reco_cv.sum():.1f}")
    print(f"  True signal (raw): {sig_true_all.sum()}")
    print(f"  Purity: {sig_reco_cv.sum()/(sig_reco_cv+bkg_reco_cv).sum()*100:.1f}%")
 
# Cross-check with syst notebook response matrix
print("\n=== Cross-check vs syst notebook CV results ===")
for v in ['reco_ke', 'reco_costheta']:
    R = response_dict[v]['response']
    print(f"{v}: shape={R.shape}, diag={np.round(np.diag(R),3)}, "
          f"mean_diag={np.mean(np.diag(R)):.4f}")
```

```python
# %% CELL 5 — Plot response matrices + efficiency
 
for var_name, vcfg in VAR_CFGS.items():
    rd = response_dict[var_name]
 
    # Response matrix heatmap
    fig, ax = plot_heatmap(rd['response'],
                            f'Response Matrix $R_{{a\\mu}}$: {var_name}',
                            vcfg, cmap='viridis')
    savefig(f'{var_name}_response_matrix.png'); plt.show()
 
    # Efficiency
    fig, ax = plt.subplots(figsize=(7, 4))
    eff = rd['efficiency']
    ax.step(vcfg.bins, np.append(eff*100, eff[-1]*100),
            where='post', color='steelblue', lw=2)
    ax.fill_between(vcfg.bins, np.append(eff*100, eff[-1]*100),
                    step='post', alpha=0.15, color='steelblue')
    ax.set_xlim(vcfg.bins[0], vcfg.bins[-1]); ax.set_ylim(0, 110)
    ax.set_xlabel(vcfg.xlabel_true); ax.set_ylabel('Efficiency [%]')
    ax.set_title(f'{TITLE}: Efficiency'); ax.grid(alpha=0.2, ls='--')
    plt.tight_layout(); savefig(f'{var_name}_efficiency.png'); plt.show()
 
    # Purity
    pur = rd['sig_reco_cv'] / (rd['sig_reco_cv'] + rd['bkg_reco_cv']).clip(1)
    fig, ax = plt.subplots(figsize=(7, 4))
    ax.step(vcfg.bins, np.append(pur*100, pur[-1]*100),
            where='post', color='crimson', lw=2)
    ax.set_xlim(vcfg.bins[0], vcfg.bins[-1]); ax.set_ylim(0, 110)
    ax.set_xlabel(vcfg.xlabel); ax.set_ylabel('Purity [%]')
    ax.set_title(f'{TITLE}: Purity'); ax.grid(alpha=0.2, ls='--')
    plt.tight_layout(); savefig(f'{var_name}_purity.png'); plt.show()
```

```python
# %% CELL 6 — Load systematic covariances
SYST_SOURCES = ['flux', 'genie', 'extra_xsec', 'g4', 'mcstat', 'pot', 'ntargets']
SRC_LABELS = {
    'flux': 'BNB Flux', 'genie': 'GENIE', 'extra_xsec': 'Extra Xsec',
    'g4': 'Geant4', 'mcstat': 'MC Stat', 'pot': 'POT',
    'ntargets': r'$N_\mathrm{targets}$', 'detsys': 'Detector',
}
 
cov_loaded = {}
for var_name in VAR_CFGS:
    cov_loaded[var_name] = load_syst_covariances(SYST_DIR, var_name, SYST_SOURCES)
    print(f"\n{var_name}: loaded {list(cov_loaded[var_name].keys())}")
 
    total_sig = build_total_covariance(cov_loaded[var_name], 'cov_ms_ms')
    if total_sig is not None:
        rd = response_dict[var_name]
        frac = np.sqrt(np.diag(total_sig).clip(0))
        frac_rel = frac / np.where(rd['sig_reco_cv'] > 0, rd['sig_reco_cv'], 1.0)
        print(f"  Total signal frac unc (per bin): {np.round(frac_rel*100, 1)}%")
        print(f"  Total signal frac unc (integrated): "
              f"{np.sqrt(total_sig.sum())/rd['sig_reco_cv'].sum()*100:.1f}%")
```

```python
# %% CELL 7 — Uncertainty budget + correlation
for var_name, vcfg in VAR_CFGS.items():
    rd = response_dict[var_name]
    cv_total = rd['sig_reco_cv'] + rd['bkg_reco_cv']
    cov_dict_plot = {SRC_LABELS.get(s, s): v['cov_ms_ms']
                     for s, v in cov_loaded[var_name].items() if 'cov_ms_ms' in v}
 
    fig, ax = plot_uncertainty_budget(vcfg, cv_total, cov_dict_plot,
                                      title=f'{TITLE}: Unc. Budget')
    savefig(f'{var_name}_uncertainty_budget.png'); plt.show()
 
    # Correlation
    total_cov = build_total_covariance(cov_loaded[var_name], 'cov_ms_ms')
    if total_cov is not None:
        dg = np.sqrt(np.diag(total_cov).clip(1e-10))
        corr = total_cov / np.outer(dg, dg)
        fig, ax = plot_heatmap(corr, f'Reco-space Correlation: {var_name}',
                                vcfg, cmap='RdBu_r')
        savefig(f'{var_name}_correlation_reco.png'); plt.show()
```

```python
# %% CELL 8 — Total covariance for unfolding
total_cov_for_unfold = {}
for var_name, vcfg in VAR_CFGS.items():
    total = build_total_covariance(cov_loaded[var_name], 'cov_ms_ms')
    if total is None:
        total = np.diag(response_dict[var_name]['sig_reco_cv'])
        print(f"  {var_name}: using Poisson stat-only covariance (no syst loaded)")
    total_cov_for_unfold[var_name] = total
```

```python
# %% CELL 9 — Wiener-SVD closure test  [FIXED]
# ══════════════════════════════════════════════════════════════════════════════
# Key: Response shape (n_reco × n_true), Model = raw truth counts,
#      Measured = Response @ Model (Asimov data), Covariance = MCstat only
 
closure_results = {}
 
for var_name, vcfg in VAR_CFGS.items():
    rd = response_dict[var_name]
    Response = rd['response']  # (n_reco × n_true)
    Model    = rd['sig_true_cv']  # raw truth counts (unweighted)
    n_reco, n_true = Response.shape
 
    # Asimov measurement: Response @ Model
    # This is the expected reco signal if truth = Model
    Measured_asimov = Response @ Model
 
    # For closure test: use MCstat-only covariance (diagonal, Poisson)
    # Lynn: "The input covariance matrix for the closure test is MC statistical error only"
    Cov_mcstat = np.diag(np.maximum(Measured_asimov, 1.0))
 
    print(f"\n{var_name}: Closure test")
    print(f"  Response: {Response.shape} (n_reco={n_reco} × n_true={n_true})")
    print(f"  Model (truth):   {Model}")
    print(f"  Measured (R@M):  {np.round(Measured_asimov, 1)}")
    print(f"  Check: R@M ≈ sig_reco_cv/pot_scale? "
          f"{np.allclose(Measured_asimov, rd['sig_reco_cv']/pot_scale, rtol=0.05)}")
 
    unfold = WienerSVD(Response, Model, Measured_asimov, Cov_mcstat,
                       C_type=2, Norm_type=0)
    closure_results[var_name] = unfold
 
    U  = unfold['unfold']
    Uc = unfold['UnfoldCov']
    Ac = unfold['AddSmear']
    ms = Ac @ Model
 
    c2 = float((U - ms) @ np.linalg.inv(Uc) @ (U - ms))
    ndf = len(U)
    pv = 1 - scipy_chi2.cdf(c2, ndf)
    print(f"  χ²/ndf = {c2:.2f}/{ndf}  (p = {pv:.3f})")
    print(f"  ← Should be ~0.0/{ndf} (p ≈ 1.0)")
 
    # Plot
    fig, ax = plot_unfolded_result(
        vcfg, unfold, model=Model, reco_signal=Measured_asimov,
        title=f'{TITLE}: Closure — {var_name}')
    savefig(f'{var_name}_closure_test.png'); plt.show()
```

```python
# %% CELL 10 — Fake data tests  [FIXED]
# ══════════════════════════════════════════════════════════════════════════════
# Lynn's approach (technote Sec 7):
#   - Add weight to signal/background events
#   - Background-subtract using CV background
#   - Unfold with total covariance (MCstat + GENIE)
#   - Compare unfolded result to smeared nominal and smeared modified truth
 
FAKE_DATA_TESTS = [
    ({0: 1.2},          "Signal ×1.2",              "sig_x1p2"),
    ({0: 0.8},          "Signal ×0.8",              "sig_x0p8"),
    ({2: 2.0, 4: 2.0},  r"$\nu_\mu$CC bkg ×2",    "numubkg_x2"),
    ({3: 2.0},           r"NC $\pi^0$ bkg ×2",     "ncpi0_x2"),
]
 
fdt_results = {}
 
for var_name, vcfg in VAR_CFGS.items():
    rd = response_dict[var_name]
    tc = TRUE_COL_MAP[var_name]
    tb = TRUE_BINS_MAP[var_name]
    Response = rd['response']
    Model    = rd['sig_true_cv']
    Cov      = total_cov_for_unfold[var_name]
 
    print(f"\n{'='*60}\nFake data tests: {var_name}\n{'='*60}")
 
    for scale_dict, label, tag in FAKE_DATA_TESTS:
        sel = sel_final.copy()
 
        # Build fake-data weights (applied to raw, unweighted events)
        wgt = np.ones(len(sel), dtype=float)
        for cat, scale in scale_dict.items():
            wgt[sel['truth_cat'].values == cat] *= scale
 
        reco_vals = sel[var_name].clip(vcfg.bins[0], vcfg.bins[-1]-_eps).values
 
        sig_m = (sel['truth_cat'] == 0).values
        bkg_m = sel['truth_cat'].between(1, 6).values
 
        # Fake reco histograms (unweighted, with fake-data scaling)
        fake_total, _ = np.histogram(reco_vals, bins=vcfg.bins, weights=wgt)
        fake_sig, _   = np.histogram(reco_vals[sig_m], bins=vcfg.bins, weights=wgt[sig_m])
        fake_bkg, _   = np.histogram(reco_vals[bkg_m], bins=vcfg.bins, weights=wgt[bkg_m])
 
        # CV background (unweighted)
        cv_bkg_raw, _ = np.histogram(reco_vals[bkg_m], bins=vcfg.bins)
 
        # Background-subtracted measurement
        Measured_fake = fake_total - cv_bkg_raw  # = fake_sig + (fake_bkg - cv_bkg)
 
        # Modified truth (for comparison after unfolding)
        all_sig = sel_topo[sel_topo['is_sig']]
        true_v = all_sig[tc].dropna().clip(tb[0], tb[-1]-_eps)
        alt_wgt = np.ones(len(true_v))
        if 0 in scale_dict:
            alt_wgt *= scale_dict[0]
        alt_truth, _ = np.histogram(true_v, bins=tb, weights=alt_wgt)
 
        # Unfold
        unfold = WienerSVD(Response, Model, Measured_fake.astype(float),
                           Cov, C_type=2, Norm_type=0)
 
        U = unfold['unfold']; Uc = unfold['UnfoldCov']; Ac = unfold['AddSmear']
        ms_nom = Ac @ Model
        ms_alt = Ac @ alt_truth.astype(float)
 
        c2_nom = float((U-ms_nom) @ np.linalg.inv(Uc) @ (U-ms_nom))
        c2_alt = float((U-ms_alt) @ np.linalg.inv(Uc) @ (U-ms_alt))
        ndf = len(U)
 
        fdt_results[(var_name, tag)] = dict(
            unfold=unfold, alt_truth=alt_truth, Measured=Measured_fake, label=label)
 
        p_nom = 1 - scipy_chi2.cdf(c2_nom, ndf)
        p_alt = 1 - scipy_chi2.cdf(c2_alt, ndf)
        print(f"  {label}:")
        print(f"    χ² vs nominal:  {c2_nom:.2f}/{ndf}  (p={p_nom:.3f})")
        print(f"    χ² vs modified: {c2_alt:.2f}/{ndf}  (p={p_alt:.3f})")
 
        # Plot
        fig, ax = plt.subplots(figsize=(8, 6))
        Uerr = np.sqrt(np.diag(Uc))
        centers = vcfg.bin_centers
        ax.errorbar(centers, U, yerr=Uerr, fmt='ko', capsize=3, ms=5,
                    label='Unfolded fake data')
        ax.step(vcfg.bins, np.append(ms_nom, ms_nom[-1]), where='post',
                color='gray', lw=1.5, ls='--',
                label=rf'$A_c \otimes$ baseline ($\chi^2$={c2_nom:.1f}/{ndf}, '
                      rf'p={p_nom:.2f})')
        ax.step(vcfg.bins, np.append(ms_alt, ms_alt[-1]), where='post',
                color='darkorange', lw=2,
                label=rf'$A_c \otimes$ {label} ($\chi^2$={c2_alt:.1f}/{ndf}, '
                      rf'p={p_alt:.2f})')
        ax.fill_between(vcfg.bins, np.append(U-Uerr, (U-Uerr)[-1]),
                        np.append(U+Uerr, (U+Uerr)[-1]),
                        step='post', alpha=0.15, color='gray')
        ax.set_xlabel(vcfg.xlabel); ax.set_ylabel('Events')
        ax.set_title(f'{TITLE}: FDT {label}')
        ax.legend(fontsize=8, frameon=False); ax.grid(alpha=0.2, ls='--')
        ax.set_xlim(vcfg.bins[0], vcfg.bins[-1])
        plt.tight_layout()
        savefig(f'fakedata/{var_name}_{tag}.png'); plt.show()
```

```python
# %% CELL 11 — Cross-section extraction (Asimov)
# ══════════════════════════════════════════════════════════════════════════════
 
N_targets, Phi = get_constants()
print(f"N_targets = {N_targets:.4e}")
print(f"Φ (flux)  = {Phi:.4e} cm⁻² / POT")
 
xsec_results = {}
 
for var_name, vcfg in VAR_CFGS.items():
    rd = response_dict[var_name]
    uf = closure_results[var_name]
 
    U  = uf['unfold']
    Uc = uf['UnfoldCov']
 
    # dσ/dx = U / (Φ · POT · N_targets · Δx)
    scale = Phi * TARGET_POT * N_targets * vcfg.bin_widths
    xsec = U / scale
    xsec_cov = Uc / np.outer(scale, scale)
    xsec_unc = np.sqrt(np.diag(xsec_cov).clip(0))
 
    truth_xsec = rd['sig_true_cv'] / scale
 
    xsec_results[var_name] = dict(
        xsec=xsec, xsec_unc=xsec_unc, xsec_cov=xsec_cov,
        truth_xsec=truth_xsec)
 
    print(f"\n{var_name}:")
    for i in range(vcfg.n_bins):
        r = xsec[i] / truth_xsec[i] if truth_xsec[i] > 0 else 0
        print(f"  bin {i}: xsec={xsec[i]:.3e}  truth={truth_xsec[i]:.3e}  "
              f"ratio={r:.3f}  unc={xsec_unc[i]:.3e}")
 
    # Plot
    fig, ax = plot_xsec_comparison(
        vcfg, xsec, xsec_unc, truth_xsec=truth_xsec,
        title=f'{TITLE}: dσ/dx (Asimov)')
    savefig(f'{var_name}_xsec_asimov.png'); plt.show()
```

```python
# %% CELL 12 — Overlay: Asimov data vs prediction (matches Lynn Fig 5)
# ══════════════════════════════════════════════════════════════════════════════
 
for var_name, vcfg in VAR_CFGS.items():
    rd = response_dict[var_name]
 
    class _VCfg:
        bins = vcfg.bins
        var_labels = vcfg.var_labels
        var_plot_name = vcfg.xlabel
 
    plot_overlay_with_cov(
        data=rd['sig_reco_cv'] + rd['bkg_reco_cv'],
        cov_data=np.diag(rd['sig_reco_cv'] + rd['bkg_reco_cv']),
        signal=rd['sig_reco_cv'],
        cov_signal=total_cov_for_unfold[var_name],
        background=rd['bkg_reco_cv'],
        cov_background=np.diag(rd['bkg_reco_cv']),
        varcfg=_VCfg(),
        title=f'Asimov test — {var_name}',
        data_label='Asimov Data (stat. unc.)',
        y_label_top='Events',
    )
    savefig(f'{var_name}_asimov_overlay.png'); plt.show()
```

```python
# %% CELL 13 — OmniFold comparison  [FIXED: both in truth space]
# ══════════════════════════════════════════════════════════════════════════════
# Previously compared reco-space (Wiener-SVD) vs truth-space (OmniFold).
# Now: propagate Wiener-SVD covariance to truth space via Cov^unf = U·Cov·U^T
# so both correlation matrices are in truth space.
 
OMNIFOLD_BINNING = {
    'true_p':        np.array([0, 200, 400, 600, 800, 1000, 1400, 2000]),
    'true_costheta': np.linspace(-1, 1, 11),
}
OF_VAR_MAP = {'reco_ke': 'true_p', 'reco_costheta': 'true_costheta'}
 
for var_name, vcfg in VAR_CFGS.items():
    of_var = OF_VAR_MAP.get(var_name)
    if of_var is None: continue
 
    of_cov_file = os.path.join(OMNIFOLD_COV, f'covariance_all_{of_var}.npz')
    if not os.path.exists(of_cov_file):
        print(f"  {var_name}: OmniFold cov not found at {of_cov_file}"); continue
 
    of_data = np.load(of_cov_file)
    of_cov = of_data['cov']
    of_mean = of_data['mean_hist']
 
    # Wiener-SVD: unfolded covariance (already in truth space)
    if var_name in closure_results:
        w_cov_truth = closure_results[var_name]['UnfoldCov']
    else:
        w_cov_truth = total_cov_for_unfold[var_name]
 
    fig, axes = plt.subplots(1, 2, figsize=(14, 6))
 
    # Wiener-SVD truth-space correlation
    w_dg = np.sqrt(np.diag(w_cov_truth).clip(1e-10))
    w_corr = w_cov_truth / np.outer(w_dg, w_dg)
    n_w = w_corr.shape[0]
    im1 = axes[0].imshow(w_corr, origin='lower', aspect='auto',
                          extent=[0, n_w, 0, n_w], vmin=-1, vmax=1, cmap='RdBu_r')
    axes[0].set_title(f'Wiener-SVD (truth space): {var_name}')
    plt.colorbar(im1, ax=axes[0])
 
    # OmniFold truth-space correlation
    of_dg = np.sqrt(np.diag(of_cov).clip(1e-10))
    of_corr = of_cov / np.outer(of_dg, of_dg)
    n_o = of_corr.shape[0]
    im2 = axes[1].imshow(of_corr, origin='lower', aspect='auto',
                          extent=[0, n_o, 0, n_o], vmin=-1, vmax=1, cmap='RdBu_r')
    axes[1].set_title(f'OmniFold (truth space): {of_var}')
    plt.colorbar(im2, ax=axes[1])
 
    plt.suptitle(f'Truth-Space Correlation Comparison: {var_name}', fontsize=14)
    plt.tight_layout()
    savefig(f'omnifold_comparison/correlation_truthspace_{var_name}.png'); plt.show()
 
    print(f"  {var_name}: Wiener-SVD {n_w}×{n_w}, OmniFold {n_o}×{n_o}")
    print(f"    W-SVD diag unc: {np.round(np.sqrt(np.diag(w_cov_truth).clip(0)), 1)}")
    print(f"    OF    diag unc: {np.round(np.sqrt(np.diag(of_cov).clip(0)), 1)}")
```

```python
# %% CELL 14 — Cross-section extraction (Asimov)
# ══════════════════════════════════════════════════════════════════════════════
 
N_targets, Phi = get_constants()
print(f"N_targets = {N_targets:.4e}")
print(f"Φ (flux)  = {Phi:.4e} cm⁻² / POT")
 
xsec_results = {}
 
for var_name, vcfg in VAR_CFGS.items():
    rd = response_dict[var_name]
    uf = closure_results[var_name]
 
    # Unfolded signal
    U = uf['unfold']
    Uc = uf['UnfoldCov']
 
    # Cross-section
    xsec = extract_differential_xsec(U, vcfg.bins, TARGET_POT,
                                      n_targets=N_targets, flux=Phi)
    # Propagate uncertainty
    scale = Phi * TARGET_POT * N_targets * np.diff(vcfg.bins)
    xsec_cov = Uc / np.outer(scale, scale)
    xsec_unc = np.sqrt(np.diag(xsec_cov).clip(0))
 
    # Truth cross-section
    truth_xsec = extract_differential_xsec(rd['sig_true_cv'], vcfg.bins,
                                            TARGET_POT, N_targets, Phi)
 
    xsec_results[var_name] = dict(
        xsec=xsec, xsec_unc=xsec_unc, xsec_cov=xsec_cov,
        truth_xsec=truth_xsec)
 
    print(f"\n{var_name}:")
    print(f"  xsec:       {xsec}")
    print(f"  xsec_unc:   {xsec_unc}")
    print(f"  truth_xsec: {truth_xsec}")
 
    # Plot
    fig, ax = plt.subplots(figsize=(6, 4))
    ax.step(vcfg.bins, np.append(truth_xsec, truth_xsec[-1]),
            where='post', color='black', lw=2, label='MC Truth')
    ax.errorbar(vcfg.bin_centers, xsec, yerr=xsec_unc,
                fmt='o', color='steelblue', capsize=3, markersize=5,
                lw=1.5, label='Wiener-SVD (Asimov)')
    ax.set_xlabel(vcfg.xlabel); ax.set_ylabel(r'd$\sigma$/d$x$')
    ax.set_title(f'{TITLE}: Differential Cross-Section')
    ax.legend(frameon=False)
    ax.ticklabel_format(axis='y', style='sci', scilimits=(-2, 2))
    ax.set_xlim(vcfg.bins[0], vcfg.bins[-1])
    plt.tight_layout()
    savefig(f'{var_name}_xsec_asimov.png'); plt.show()
```

```python
# %% CELL 15 — Save covariance matrices for downstream use
# ══════════════════════════════════════════════════════════════════════════════
 
for var_name, vcfg in VAR_CFGS.items():
    rd = response_dict[var_name]

    cov_bs = build_total_covariance(cov_loaded[var_name], 'cov_bs_bs')
    if cov_bs is None:
        cov_bs = np.zeros((vcfg.n_bins, vcfg.n_bins))

    save_cov_npz(
        os.path.join(COV_OUT_DIR, f'{var_name}_all_cov_matrices.npz'),
        response=rd['response'],
        true_signal=rd['sig_true_cv'],
        measured_signal=rd['sig_reco_cv'],
        measured_background=rd['bkg_reco_cv'],
        cov_ms_ms={'cov': total_cov_for_unfold[var_name]},
        cov_bs_bs={'cov': cov_bs},
    )
```

```python
# %% CELL 16 — OmniFold comparison
# ══════════════════════════════════════════════════════════════════════════════
# Load OmniFold results and compare with Wiener-SVD
 
print("=== OmniFold Comparison ===")
 
OMNIFOLD_BINNING = {
    'true_p':        np.array([0, 200, 400, 600, 800, 1000, 1400, 2000]),
    'true_costheta': np.linspace(-1, 1, 11),
}
 
# Map our variable names to OmniFold variable names
OF_VAR_MAP = {
    'reco_ke': 'true_p',            # OmniFold uses true_p (momentum)
    'reco_costheta': 'true_costheta',
}
 
for var_name, vcfg in VAR_CFGS.items():
    of_var = OF_VAR_MAP.get(var_name)
    if of_var is None: continue
 
    # Check if OmniFold covariance exists
    of_cov_file = os.path.join(OMNIFOLD_COV, f'covariance_all_{of_var}.npz')
    if not os.path.exists(of_cov_file):
        print(f"  {var_name}: OmniFold cov not found at {of_cov_file}")
        continue
 
    of_data = np.load(of_cov_file)
    of_cov = of_data['cov']
    of_mean = of_data['mean_hist']
    of_bins = OMNIFOLD_BINNING[of_var]
 
    # Plot correlation comparison
    fig, axes = plt.subplots(1, 2, figsize=(14, 6))
 
    # Wiener-SVD correlation
    w_cov = total_cov_for_unfold[var_name]
    w_dg = np.sqrt(np.diag(w_cov).clip(1e-10))
    w_corr = w_cov / np.outer(w_dg, w_dg)
 
    im1 = axes[0].imshow(w_corr, origin='lower', aspect='auto',
                          vmin=-1, vmax=1, cmap='RdBu_r')
    axes[0].set_title(f'Wiener-SVD: {var_name}')
    plt.colorbar(im1, ax=axes[0])
 
    # OmniFold correlation
    of_dg = np.sqrt(np.diag(of_cov).clip(1e-10))
    of_corr = of_cov / np.outer(of_dg, of_dg)
    im2 = axes[1].imshow(of_corr, origin='lower', aspect='auto',
                          vmin=-1, vmax=1, cmap='RdBu_r')
    axes[1].set_title(f'OmniFold: {of_var}')
    plt.colorbar(im2, ax=axes[1])
 
    plt.suptitle(f'Correlation Matrix Comparison: {var_name}')
    plt.tight_layout()
    savefig(f'omnifold_comparison/correlation_{var_name}.png'); plt.show()
 
    # Uncertainty budget comparison
    fig, axes = plt.subplots(1, 2, figsize=(13, 5))
    for src in ['bnb', 'genie', 'extra_xsec', 'g4', 'mcstat']:
        src_file = os.path.join(OMNIFOLD_COV, f'covariance_{src}_{of_var}.npz')
        if not os.path.exists(src_file): continue
        sd = np.load(src_file)
        frac_of = np.sqrt(np.diag(sd['cov']).clip(0)) / sd['mean_hist'].clip(1e-6)
        axes[1].step(of_bins, np.append(frac_of, frac_of[-1]),
                     where='post', lw=1.5, label=src)
    frac_of_total = np.sqrt(np.diag(of_cov).clip(0)) / of_mean.clip(1e-6)
    axes[1].step(of_bins, np.append(frac_of_total, frac_of_total[-1]),
                 where='post', color='black', lw=2, ls='--', label='Total')
    axes[1].set_xlabel(f'{of_var}'); axes[1].set_ylabel('Frac. unc.')
    axes[1].set_title('OmniFold'); axes[1].legend(fontsize=8)
 
    # Wiener-SVD
    for src, cov_entry in cov_loaded[var_name].items():
        if 'cov_ms_ms' not in cov_entry: continue
        frac_w = (np.sqrt(np.diag(cov_entry['cov_ms_ms']).clip(0))
                  / response_dict[var_name]['sig_reco_cv'].clip(1e-6))
        axes[0].step(vcfg.bins, np.append(frac_w, frac_w[-1]),
                     where='post', lw=1.5, label=SRC_LABELS.get(src, src))
    frac_w_total = (np.sqrt(np.diag(w_cov).clip(0))
                    / response_dict[var_name]['sig_reco_cv'].clip(1e-6))
    axes[0].step(vcfg.bins, np.append(frac_w_total, frac_w_total[-1]),
                 where='post', color='black', lw=2, ls='--', label='Total')
    axes[0].set_xlabel(vcfg.xlabel); axes[0].set_ylabel('Frac. unc.')
    axes[0].set_title('Wiener-SVD'); axes[0].legend(fontsize=8)
 
    plt.suptitle(f'Uncertainty Budget Comparison: {var_name}')
    plt.tight_layout()
    savefig(f'omnifold_comparison/unc_budget_{var_name}.png'); plt.show()
 
print("\nOmniFold comparison plots saved.")
```

```python
# %% CELL 18 — Purity and efficiency per bin (all variables)
# ══════════════════════════════════════════════════════════════════════════════
 
for var_name, vcfg in VAR_CFGS.items():
    rd = response_dict[var_name]
    purity = rd['sig_reco_cv'] / (rd['sig_reco_cv'] + rd['bkg_reco_cv']).clip(1)
    eff = rd['efficiency']
 
    fig, axes = plt.subplots(1, 2, figsize=(12, 5))
    for ax, vals, ylabel, color, glbl in [
        (axes[0], purity*100, 'Purity [%]', 'steelblue',
         f"Global: {rd['sig_reco_cv'].sum()/(rd['sig_reco_cv']+rd['bkg_reco_cv']).sum()*100:.1f}%"),
        (axes[1], eff*100, 'Efficiency [%]', 'crimson',
         f"Global: {rd['sig_true_sel'].sum()/rd['sig_true_cv'].sum()*100:.1f}%"),
    ]:
        ax.step(vcfg.bins, np.append(vals, vals[-1]),
                where='post', color=color, lw=2)
        ax.fill_between(vcfg.bins, np.append(vals, vals[-1]),
                        step='post', alpha=0.15, color=color)
        ax.set_xlim(vcfg.bins[0], vcfg.bins[-1])
        ax.set_ylim(0, 110)
        ax.set_xlabel(vcfg.xlabel, fontsize=12)
        ax.set_ylabel(ylabel, fontsize=12)
        ax.grid(alpha=0.2, ls='--')
        ax.text(0.05, 0.95, glbl, transform=ax.transAxes, va='top',
                fontsize=11, bbox=dict(fc='white', alpha=0.6, ec='none'))
 
    axes[0].set_title(f'{TITLE}: Signal Purity')
    axes[1].set_title(f'{TITLE}: Signal Efficiency')
    plt.tight_layout()
    savefig(f'{var_name}_purity_efficiency.png'); plt.show()
```

```python
# %% CELL 19 — Summary
# ══════════════════════════════════════════════════════════════════════════════
 
print(f"\n{'='*60}")
print(f"Cross-Section Analysis Summary")
print(f"{'='*60}")
print(f"Target POT:    {TARGET_POT:.2e}")
print(f"N_targets:     {N_targets:.4e}")
print(f"Flux:          {Phi:.4e} cm⁻² / POT")
print(f"Final stage:   {FINAL_STAGE}")
print(f"")
 
for var_name, vcfg in VAR_CFGS.items():
    rd = response_dict[var_name]
    total_cov = total_cov_for_unfold[var_name]
    total_unc = np.sqrt(np.diag(total_cov).sum()) / rd['sig_reco_cv'].sum()
 
    print(f"\n{var_name}:")
    print(f"  Bins:         {vcfg.n_bins}")
    print(f"  Selected sig: {rd['sig_reco_cv'].sum():.0f} (weighted)")
    print(f"  Selected bkg: {rd['bkg_reco_cv'].sum():.0f} (weighted)")
    print(f"  Purity:       {rd['sig_reco_cv'].sum()/(rd['sig_reco_cv']+rd['bkg_reco_cv']).sum()*100:.1f}%")
    print(f"  Mean eff:     {rd['efficiency'].mean()*100:.1f}%")
    print(f"  Total unc:    {total_unc*100:.1f}%")
 
    if var_name in closure_results:
        uf = closure_results[var_name]
        U = uf['unfold']; Uc = uf['UnfoldCov']; Ac = uf['AddSmear']
        ms = Ac @ rd['sig_true_cv']
        c2 = float((U-ms) @ np.linalg.inv(Uc) @ (U-ms))
        print(f"  Closure χ²:   {c2:.2f}/{len(U)}")
 
print(f"\n{'='*60}")
print(f"Output directories:")
print(f"  Plots:    {PLOT_DIR}/")
print(f"  Cov NPZ:  {COV_OUT_DIR}/")
print(f"  Syst in:  {SYST_DIR}/")
print(f"{'='*60}")
```
