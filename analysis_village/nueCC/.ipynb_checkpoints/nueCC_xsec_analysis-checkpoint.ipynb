```python
# CELL 4 — POT normalisation & physical constants

# ── Total MC POT from histpotdf ───────────────────────────────────────────────
hist_frames = []
for k in range(n_splits):
    try:
        hist_frames.append(pd.read_hdf(DF_FILE, key=f"histpotdf_{k}"))
    except KeyError:
        pass
histpotdf = pd.concat(hist_frames, axis=0) if hist_frames else pd.DataFrame()

pot_col    = "TotalPOT" if "TotalPOT" in histpotdf.columns \
             else histpotdf.select_dtypes("number").columns[0]
mc_tot_pot = float(histpotdf[pot_col].sum())

if mc_tot_pot == 0:
    raise RuntimeError("mc_tot_pot=0 — check histpotdf")

TARGET_POT = 6.6e20
pot_scale  = TARGET_POT / mc_tot_pot

print(f"MC total POT  : {mc_tot_pot:.3e}")
print(f"Target POT    : {TARGET_POT:.3e}")
print(f"POT scale     : {pot_scale:.4f}")

# ── Physical constants ─────────────────────────────────────────────────────────
RHO_AR    = 1.3836           # g/cm³
N_A       = 6.02214076e23
M_AR      = 39.948           # g/mol
V_ACTIVE  = 2 * 200 * 400 * 500   # cm³  ← update with official FV definition
N_TARGETS = RHO_AR * V_ACTIVE * N_A / M_AR
print(f"N_targets     : {N_TARGETS:.3e}")

# ── Integrated nue flux ───────────────────────────────────────────────────────
# TODO: load from ROOT flux file once path is confirmed:
#   import uproot
#   flux     = uproot.open("/exp/sbnd/data/users/.../sbnd_flux.root")
#   nue_flux = flux["flux_sbnd_nue"].to_numpy()           # (values, edges)
#   INTEGRATED_FLUX = nue_flux[0].sum() / 1e4             # /m² → /cm²
#   INTEGRATED_FLUX *= mc_tot_pot / 1e6                   # per 10^6 POT → per mc_tot_pot POT
#
INTEGRATED_FLUX  = 1.0    # PLACEHOLDER — update with real nue flux
XSEC_UNIT        = 1.0 / (INTEGRATED_FLUX * N_TARGETS * TARGET_POT)
XSEC_UNIT_EVENT  = 1.0   # dimensionless; use for event-rate closure tests

print(f"\nWARNING: INTEGRATED_FLUX is a placeholder — cross-section units not yet valid")
print(f"XSEC_UNIT        : {XSEC_UNIT:.3e}  (placeholder)")
print(f"XSEC_UNIT_EVENT  : {XSEC_UNIT_EVENT}  (event-rate mode, flux-independent)")
```

```python
# CELL 15 — Response matrix

# Selected signal: reco and truth variable values (from sel_qual)
sel_sig_df      = sel_df[sig_mask_sel]
var_reco_sel_sig = np.clip(sel_sig_df[RECO_COL].values, eps_b, eps_e)
var_true_sel_sig = np.clip(sel_sig_df[TRUE_COL].values, eps_b, eps_e)
weight_sel_sig   = np.full(len(sel_sig_df), pot_scale)

# ── Smearing matrix ────────────────────────────────────────────────────────────
fig_name = os.path.join(PLOT_DIR, f"{VAR_NAME}_smearing_matrix")
reco_vs_true = get_smear_matrix(
    var_true_sel_sig, var_reco_sel_sig, bins,
    var_labels=[VAR_LABEL, VAR_LABEL_RECO, VAR_LABEL_TRUE],
    save_fig=SAVE_FIGS, save_fig_name=fig_name,
)
plt.show()

# ── Efficiency vector ──────────────────────────────────────────────────────────
eff_vec = get_eff(reco_vs_true, nevts_signal_truth)

fig, ax = plt.subplots(figsize=(7, 4))
ax.step(bins, np.append(eff_vec*100, eff_vec[-1]*100),
        where="post", color="steelblue", lw=2)
ax.fill_between(bins, np.append(eff_vec*100, eff_vec[-1]*100),
                step="post", alpha=0.15, color="steelblue")
ax.set_xlim(bins[0], bins[-1]); ax.set_ylim(0, 110)
ax.set_xlabel(VAR_LABEL_TRUE, fontsize=12)
ax.set_ylabel("Efficiency [%]", fontsize=12)
ax.set_title(f"{TITLE_PREFIX}: Efficiency Vector")
ax.grid(alpha=0.2, ls="--")
plt.tight_layout()
savefig(f"{VAR_NAME}_efficiency.png")
plt.show()

# ── Response matrix ────────────────────────────────────────────────────────────
fig_name = os.path.join(PLOT_DIR, f"{VAR_NAME}_response_matrix")
Response = get_response_matrix(
    reco_vs_true, eff_vec, bins,
    var_labels=[VAR_LABEL, VAR_LABEL_RECO, VAR_LABEL_TRUE],
    save_fig=SAVE_FIGS, save_fig_name=fig_name,
)
print("Response matrix:")
print(np.round(Response, 3))
print(f"\nColumn sums (≈ efficiency): {np.round(Response.sum(axis=0), 3)}")
plt.show()
```

```python
# CELL 16 — MCstat Poisson universes (flat df version)
# sel_qual is one row per interaction → straightforward Poisson resampling.
# Seeds are derived from the df integer index for reproducibility.

rng = np.random.default_rng(42)
N   = len(sel_df)   # only selected interactions need universe weights

# Deterministic per-event seeds from row position in sel_df
row_seeds    = np.arange(N, dtype=np.uint32)
univ_seeds   = rng.integers(0, 2**31, size=N_UNIVERSES, dtype=np.uint32)

# Pre-draw all Poisson weights at once: (N_UNIVERSES × N) matrix
# Each column is one event, each row is one universe.
# Combined seed = event_seed XOR universe_seed (cheap, bijective)
mcstat_weights = np.ones((N_UNIVERSES, N), dtype=np.float32)
for u in tqdm(range(N_UNIVERSES), desc="Drawing MCstat universes"):
    combined = (row_seeds + univ_seeds[u]) % (2**32)
    rng_u    = np.random.default_rng(int(combined.sum()) % (2**32))
    # Vectorised: one Poisson draw per event
    mcstat_weights[u] = rng_u.poisson(1.0, size=N).astype(np.float32)

# Sanity: mean of universe weights ≈ 1.0
print(f"Mean universe weight (expect ≈ 1.0): {mcstat_weights.mean():.4f}")

# For convenience, store in a DataFrame aligned with sel_df
mcstat_df = pd.DataFrame(
    mcstat_weights.T,
    index   = sel_df.index,
    columns = [f"univ_{u}" for u in range(N_UNIVERSES)],
)
```

```python
# CELL 17 — MCstat universe scatter plot

fig, axes = plt.subplots(2, 1, figsize=(8, 7),
                          gridspec_kw={"height_ratios": [3, 1]}, sharex=True)

cv_hist = nevts_signal_sel_reco   # already computed

sig_vals = np.clip(sel_df.loc[sig_mask_sel, RECO_COL].values, eps_b, eps_e)

for u in range(min(N_UNIVERSES, 60)):
    wgt_u   = mcstat_df.loc[sig_mask_sel, f"univ_{u}"].values * pot_scale
    uh, _   = np.histogram(sig_vals, bins=bins, weights=wgt_u)
    axes[0].step(bins, np.append(uh, uh[-1]), where="post",
                 color="gray", alpha=0.3, lw=0.6)

axes[0].step(bins, np.append(cv_hist, cv_hist[-1]),
             where="post", color="black", lw=2, label="CV")
axes[0].set_ylabel("Events", fontsize=12)
axes[0].set_title(f"{TITLE_PREFIX}: MCstat Universe Scatter ({N_UNIVERSES} Poisson)")
axes[0].legend(frameon=False)
axes[0].grid(alpha=0.2, ls="--")

# Ratio: universe mean / CV
univ_stack = np.stack([
    np.histogram(sig_vals, bins=bins,
                 weights=mcstat_df.loc[sig_mask_sel, f"univ_{u}"].values * pot_scale)[0]
    for u in range(N_UNIVERSES)
])
ratio_mean = np.divide(univ_stack.mean(axis=0), cv_hist,
                       out=np.ones(len(cv_hist), dtype=float),
                       where=cv_hist > 0)
axes[1].step(bins, np.append(ratio_mean, ratio_mean[-1]),
             where="post", color="black", lw=2)
axes[1].axhline(1.0, color="red", ls="--", lw=1)
axes[1].set_ylim(0.9, 1.1)
axes[1].set_xlabel(VAR_LABEL_RECO, fontsize=12)
axes[1].set_ylabel("Univ. mean / CV", fontsize=11)
axes[1].grid(alpha=0.2, ls="--")

plt.tight_layout()
savefig(f"{VAR_NAME}_mcstat_universes.png")
plt.show()
```

```python
# CELL 18 — MCstat covariance matrix

n_bins   = len(bins) - 1
Cov      = np.zeros((n_bins, n_bins))
CovFrac  = np.zeros((n_bins, n_bins))

sig_vals_clip = np.clip(sel_df.loc[sig_mask_sel, RECO_COL].values, eps_b, eps_e)
bkg_vals_clip = np.clip(sel_df.loc[bkg_mask_sel, RECO_COL].values, eps_b, eps_e)

cv_sig = nevts_signal_sel_reco * XSEC_UNIT_EVENT
cv_bkg = nevts_bkg_sel_reco    * XSEC_UNIT_EVENT

for u in tqdm(range(N_UNIVERSES), desc="MCstat covariance"):
    wgt_sig_u = mcstat_df.loc[sig_mask_sel, f"univ_{u}"].values * pot_scale
    wgt_bkg_u = mcstat_df.loc[bkg_mask_sel, f"univ_{u}"].values * pot_scale

    sig_u, _ = np.histogram(sig_vals_clip, bins=bins, weights=wgt_sig_u)
    bkg_u, _ = np.histogram(bkg_vals_clip, bins=bins, weights=wgt_bkg_u)

    # Universe variation in (signal – background) = signal contribution only
    # Background variation enters via delta_bkg = bkg_univ – bkg_cv
    delta_bkg = (bkg_u - nevts_bkg_sel_reco) * XSEC_UNIT_EVENT
    univ_val  = sig_u * XSEC_UNIT_EVENT + delta_bkg   # total universe

    dv  = univ_val - cv_sig
    with np.errstate(divide="ignore", invalid="ignore"):
        dvf = np.where(cv_sig != 0, dv / cv_sig, 0.0)

    Cov     += np.outer(dv,  dv)
    CovFrac += np.outer(dvf, dvf)

Cov     /= N_UNIVERSES
CovFrac /= N_UNIVERSES

diag_r  = np.sqrt(np.diag(Cov))
with np.errstate(divide="ignore", invalid="ignore"):
    Corr = Cov / np.outer(np.where(diag_r > 0, diag_r, 1),
                           np.where(diag_r > 0, diag_r, 1))

ret_mcstat = dict(Covariance=Cov, Covariance_Frac=CovFrac, Correlation=Corr)

print("Fractional MCstat uncertainty per bin:")
for i, (lo, hi, f) in enumerate(zip(bins[:-1], bins[1:],
                                     np.sqrt(np.diag(CovFrac)))):
    print(f"  [{lo:.0f}, {hi:.0f}] MeV  →  {f*100:.1f}%")
```

```python
# CELL 24 — Fake data test driver

def run_fake_data_test(scale_dict, label,
                       sel_df, sel_mask_sig, sel_mask_bkg,
                       mcstat_df, bins, eps_b, eps_e, pot_scale,
                       Response, Model, Cov,
                       presel_sig_df,   # sel_qual[truth_cat==0] for alt truth
                       XSEC_UNIT_EVENT, C_type=2, Norm_type=0):
    """
    scale_dict : {truth_cat: scale}  applied to selected events of that category.
    presel_sig_df : slice of sel_qual with truth_cat==0 (for alternative truth).
    """
    # Build fake-data event weights
    wgt_fake = np.full(len(sel_df), pot_scale, dtype=float)
    for cat, scale in scale_dict.items():
        wgt_fake[sel_df["truth_cat"].values == cat] *= scale

    fake_sig_vals = np.clip(sel_df.loc[sel_mask_sig, RECO_COL].values, eps_b, eps_e)
    fake_bkg_vals = np.clip(sel_df.loc[sel_mask_bkg, RECO_COL].values, eps_b, eps_e)
    fake_sig_wgt  = wgt_fake[sel_mask_sig.values]
    fake_bkg_wgt  = wgt_fake[sel_mask_bkg.values]

    fake_sig_hist, _ = np.histogram(fake_sig_vals, bins=bins, weights=fake_sig_wgt)
    fake_bkg_hist, _ = np.histogram(fake_bkg_vals, bins=bins, weights=fake_bkg_wgt)
    Measured_fake = (fake_sig_hist - (fake_bkg_hist - nevts_bkg_sel_reco)) * XSEC_UNIT_EVENT

    unfold_fake = WienerSVD(Response, Model, Measured_fake, Cov, C_type, Norm_type)

    # Alternative truth (only signal reweighting changes the truth spectrum)
    alt_truth = nevts_signal_truth.copy().astype(float)
    if 0 in scale_dict:
        true_vals = np.clip(presel_sig_df[TRUE_COL].values, eps_b, eps_e)
        cv_b, _  = np.histogram(true_vals, bins=bins)
        alt_b, _ = np.histogram(true_vals, bins=bins,
                                weights=np.full(len(true_vals), scale_dict[0]))
        alt_truth += (alt_b - cv_b)

    U    = unfold_fake["unfold"]
    Uc   = unfold_fake["UnfoldCov"]
    Uerr = np.sqrt(np.diag(Uc))
    Ac   = unfold_fake["AddSmear"]

    model_smear     = Ac @ Model
    alt_smear       = Ac @ (alt_truth * XSEC_UNIT_EVENT)

    def _chi2(a, b, C):
        return float((a - b) @ np.linalg.inv(C) @ (a - b))

    c2_nom = _chi2(U, model_smear, Uc)
    c2_alt = _chi2(U, alt_smear,   Uc)
    ndf    = len(U)

    fig, ax = plt.subplots(figsize=(9, 6))
    ax.step(bins, np.append(U, U[-1]), where="post", color="black", lw=2, label="Unfolded")
    ax.bar(bin_centers, 2*Uerr, width=np.diff(bins),
           bottom=U - Uerr, color="gray", alpha=0.4, lw=0, label="Unfolded unc.")
    ax.plot(bin_centers, Measured_fake, "o", color="steelblue", ms=5,
            label="Reco. signal (fake data)")
    ax.plot(bin_centers, model_smear, "s", color="crimson", ms=5,
            label=r"$A_c\times$ Nominal truth "
                  r"($\chi^2/\mathrm{ndf}=%.2f/%d$)" % (c2_nom, ndf))
    ax.plot(bin_centers, alt_smear, "D", color="darkorange", ms=5,
            label=r"$A_c\times$ " + label + r" truth "
                  r"($\chi^2/\mathrm{ndf}=%.2f/%d$)" % (c2_alt, ndf))
    ax.set_xlim(bins[0], bins[-1])
    ax.set_xlabel(VAR_LABEL, fontsize=12)
    ax.set_ylabel("Events", fontsize=12)
    ax.set_title(f"{TITLE_PREFIX}: Fake Data — {label}", fontsize=12)
    ax.legend(frameon=False, fontsize=9)
    ax.grid(alpha=0.2, ls="--")
    plt.tight_layout()
    return fig, ax, unfold_fake

presel_sig_df = sel_qual[presel_sig_mask]   # truth_cat==0, all FM+FV signal
```

```python
# CELL 25 — Fake data: signal × 1.2
fig, ax, uf_sig12 = run_fake_data_test(
    {0: 1.2}, "Signal ×1.2",
    sel_df, sig_mask_sel, bkg_mask_sel, mcstat_df,
    bins, eps_b, eps_e, pot_scale,
    Response, Model, Cov_tot, presel_sig_df, XSEC_UNIT_EVENT,
)
savefig(f"{VAR_NAME}_fake_sig_x1p2.png"); plt.show()

# Fake data: signal × 0.8
fig, ax, uf_sig08 = run_fake_data_test(
    {0: 0.8}, "Signal ×0.8",
    sel_df, sig_mask_sel, bkg_mask_sel, mcstat_df,
    bins, eps_b, eps_e, pot_scale,
    Response, Model, Cov_tot, presel_sig_df, XSEC_UNIT_EVENT,
)
savefig(f"{VAR_NAME}_fake_sig_x0p8.png"); plt.show()

# Fake data: numuCC background (cat 2+4) × 2
fig, ax, uf_bkg2 = run_fake_data_test(
    {2: 2.0, 4: 2.0}, r"$\nu_\mu$CC bkg ×2",
    sel_df, sig_mask_sel, bkg_mask_sel, mcstat_df,
    bins, eps_b, eps_e, pot_scale,
    Response, Model, Cov_tot, presel_sig_df, XSEC_UNIT_EVENT,
)
savefig(f"{VAR_NAME}_fake_numubkg_x2.png"); plt.show()

# Fake data: NC pi0 (cat 3) × 2
fig, ax, uf_ncpi0 = run_fake_data_test(
    {3: 2.0}, r"NC $\pi^0$ bkg ×2",
    sel_df, sig_mask_sel, bkg_mask_sel, mcstat_df,
    bins, eps_b, eps_e, pot_scale,
    Response, Model, Cov_tot, presel_sig_df, XSEC_UNIT_EVENT,
)
savefig(f"{VAR_NAME}_fake_ncpi0_x2.png"); plt.show()
```

```python
# CELL 26 — χ²/ndf summary table

print(f"\n{'Test':<35}  chi2    ndf   p-val")
print("-" * 58)
for label, unfold_res in [
    ("Closure (Asimov)",        unfold_closure),
    ("Fake: signal ×1.2",       uf_sig12),
    ("Fake: signal ×0.8",       uf_sig08),
    ("Fake: numuCC bkg ×2",     uf_bkg2),
    ("Fake: NC pi0 bkg ×2",     uf_ncpi0),
]:
    U  = unfold_res["unfold"]
    Uc = unfold_res["UnfoldCov"]
    Ms = unfold_res["AddSmear"] @ Model
    c2 = float((U - Ms) @ np.linalg.inv(Uc) @ (U - Ms))
    ndf = len(U)
    pv  = 1 - scipy_chi2.cdf(c2, ndf)
    print(f"  {label:<33}  {c2:6.2f}   {ndf}   {pv:.3f}")
```
